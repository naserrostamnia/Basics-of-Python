# Library Management System using OOP in Python

## Objective
Design and implement a library system that models different types of people (**students, staff, and visitors**) who can borrow books under specific rules. The system should demonstrate key OOP principles such as `inheritance`, `encapsulation`, `polymorphism`, and the use of `static and class methods`.

---

## Class Structure

### 1. Base Class: `Person`
Represents a generic library user.

| Attribute             | Type          | Visibility  | Description                    |
|-----------------------|---------------|-------------|-------------------------------|
| `__name`              | `str`         | Private     | Person's name                 |
| `__id_number`         | `str`         | Private     | Unique identifier             |
| `__borrowed_books`    | `list[str]`   | Private     | List of borrowed book titles |
| `_max_books_allowed`  | `int`         | Protected   | Borrow limit set by subclass |
| `total_people`        | `int`         | Class-level | Tracks number of users        |

**Key Methods:**
- `borrow_book(book: str, available_books: list[str])`
- `return_book(book: str, available_books: list[str])`
- `show_info()`
- `get_total_people()` (class method)
- `library_rules()` (static method)
- `__repr__()`

---

## Features Demonstrated
- Object-Oriented Programming principles:
  - Inheritance
  - Encapsulation
  - Polymorphism
- Data validation via assertions
- Class vs instance methods
- Use of private attributes with property getters/setters
- Static methods for general library rules

---

```python
class Person:
  total_people = 0  # Class attribute

  def __init__(self, name:str, id_number:str):
    assert isinstance(name, str) and name.strip(), "Name must be a non-empty string"
    assert isinstance(id_number, str) and id_number.strip(), "ID must be a non-empty string"

    self.__name = name
    self.__id_number = id_number
    self.__borrowed_books = []
    Person.total_people += 1
#_____________________________ Getter and Setter _____________________
  @property
  def name(self):
    return self.__name

  @name.setter
  def name(self, new_name):
    assert isinstance(new_name, str) and new_name.strip(), "Name must be a non-empty string"
    self.__name =  new_name
#______________________________
  @property
  def id_number(self):
    return self.__id_number

  @id_number.setter
  def id_number(self, value):
    self.__id_number = value
#______________________________
  @property
  def borrowed_books(self):
    return self.__borrowed_books

  @borrowed_books.setter
  def borrowed_books(self, value):
    self.__borrowed_books = value

#_____________________________Methods:_______________________________________

  def __repr__(self):
    return (f"{self.__class__.__name__}(name='{self.__name}', "
            f"id_number='{self.__id_number}', "
            f"borrowed_books={self.__borrowed_books}, "
            f"max_books_allowed={self._max_books_allowed})")

#______________________________
  def borrow_book(self, book, available_books):
    if book not in available_books:
      print(f"Sorry, {self.__name}! '{book}' is currently not available.'")
      return

    if len(self.__borrowed_books) < self._max_books_allowed:
      self.borrowed_books.append(book)
      available_books.remove(book)
      print(f"{self.__name} borrowed '{book}'.")
    else:
      print(f"{self.__name} cannot borrow more than {self._max_books_allowed} books.")

#______________________________
  def return_book(self, book, available_books):
    if book in self.__borrowed_books:
      self.__borrowed_books.remove(book)
      available_books.append(book)
      print(f"{self.__name} returned '{book}'")
    else:
          print(f"{self.__name} does not have '{book}' borrowed.")

#______________________________
  def show_info(self):
    print("\n________Information________")
    print(f"Name: {self.__name.title()}"
          f" \nID: {self.__id_number}"
          f" \nBorrowed books: {self.__borrowed_books}")

#____________________________ Class and Static Method ______________________
  @classmethod
  def get_total_people(cls):
      return cls.total_people

  @staticmethod
  def library_rules():
      return "Return books within 2 weeks. Max books vary by person type."
```

### 2. Derived Class: `Student` (inherits from `Person`)

| Attribute            | Type      | Visibility | Description           |
|----------------------|-----------|------------|-----------------------|
| `__grade_level`      | `int`     | Private    | Student’s grade (1–12)|
| `_max_books_allowed` | `int`     | Protected  | Set to 3              |

Includes validation for `grade_level` to be between 1 and 12.

---

```python
############################## Child Class: Student ############################
class Student(Person):
  def __init__(self,name:str, id_number:str, grade_level:int) -> None:
    super().__init__(name, id_number)

    self.__grade_level = grade_level
    self._max_books_allowed = 3

  @property
  def grade_level(self):
    return self.__grade_level

  @grade_level.setter
  def grade_level(self, value):
    assert isinstance(value, int) and 1 <= value <= 12, "Grade must be between 1 and 12"
    self.__grade_level = value

  def show_info(self):
    super().show_info()
    print(f"Grade Level: {self.__grade_level}")
```

### 3. Derived Class: `Staff` (inherits from `Person`)

| Attribute            | Type      | Visibility | Description          |
|----------------------|-----------|------------|----------------------|
| `__department`       | `str`     | Private    | Staff department     |
| `_max_books_allowed` | `int`     | Protected  | Set to 5             |

---


```python
############################## Child Class: Staff ##############################
class Staff(Person):
  def __init__(self, name:str, id_number:str, department:str):
    super().__init__(name, id_number)

    assert isinstance(department, str) and department.strip(), "Department must be a non-empty string"

    self.__department = department
    self._max_books_allowed = 5

  @property
  def department(self):
    return self.__department

  @department.setter
  def department(self, value):
    assert isinstance(value, str) and value.strip(), "Department must be a non-empty string"

    self.department = value


  def show_info(self):
    super().show_info()
    print(f"Department: {self.__department}")
```


### 4. Derived Class: `Visitor` (inherits from `Person`)

| Attribute             | Type      | Visibility | Description          |
|-----------------------|-----------|------------|----------------------|
| `__visit_purpose`     | `str`     | Private    | Reason for visiting  |
| `_max_books_allowed`  | `int`     | Protected  | Set to 1             |

---

```python
############################## Child Class: Visitor ##############################
class Visitor(Person):
  def __init__(self, name:str, id_number:str, visit_purpose:str):
    super().__init__(name, id_number)

    assert isinstance(visit_purpose, str) and visit_purpose.strip(), "Department must be a non-empty string"

    self.__visit_purpose = visit_purpose
    self._max_books_allowed = 1

  @property
  def visit_purpose(self):
    return self.__visit_purpose

  @visit_purpose.setter
  def visit_purpose(self, value):
    assert isinstance(value, str) and value.strip(), "visit purpose must be a non-empty string"
    self.visit_purpose = value


  def show_info(self):
    super().show_info()
    print(f"Visit purpose: {self.__visit_purpose}")
```
