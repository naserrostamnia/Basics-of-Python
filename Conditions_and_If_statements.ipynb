# Python Conditions and If statements


Python supports the usual logical conditions from mathematics:

* Equals: a == b
* Not Equals: a != b
* Less than: a < b
* Less than or equal to: a <= b
* Greater than: a > b
* Greater than or equal to: a >= b

These conditions can be used in several ways, most commonly in "if statements" and loops.

**If Statement Syntax**:
An if statement is written using the `if` keyword, **followed by a condition** and a colon `:`. **The code inside the if statement must be indented**.

```python
ages = [16, 27, 24, 17, 38]

for age in ages:
  if age >= 18:

    """ The number in \033[4m is called an ANSI escape code, and different numbers represent different text styles.
        \033[0m → Reset (removes all formatting)
        \033[1m → Bold
        \033[3m → Italic (not supported in all terminals)
        \033[4m → Underline
        \033[7m → Inverted colors (swap foreground and background)
        \033[9m → Strikethrough (not always supported)"""

    print(f"you are {age} years old and you are eligible to vote!")
  else :
    print(f"you are \33[4m {age} \33[0m years old and you \033[4mare not\033[0m eligible to vote!")

```

```python
ages_dic = {
    "Tom" : 15,
    "John" : 19,
    "Mia" : 23,
    "Hanna" : 16
}

for name, age in ages_dic.items():
  if age >= 18:
     print(f"Great news {name}! you are {age} years old and you are eligible to vote!")
  else :
    print(f"Sorry {name}! you are \33[4m {age} \33[0m years old and you \033[4mare not\033[0m eligible to vote!")
```

**if-elif-else Statement**
The `if-elif-else` statement allows you to **check multiple conditions** and execute different blocks of code depending on which condition is `True`.

```python
"""
This program determines the admission fee based on a person's age.
- Children under 6 get free entry.
- Anyone 18 or younger is eligible for a one-year free subscription.
- Adults under 65 pay $15.
- Seniors (65 and above) pay $5.
"""

ages=[3, 17, 18, 60, 65]

for age in ages:
    if age < 6:
        print(f"your child is too young to pay any admition fee.")
    elif age <= 18 :
        print(f"you are [{age}] years old and you are eligible for one-year free subscription.")
    elif age < 65  :
        print(f"you are [{age}] years old and your admition fee is $ 15.")
    else:
        print(f"you are [{age}] years old and your admition fee is $ 5.")
```

**One-Line if-else Statement**

Note: Best for short and simple conditions.

```python
"""
<expression_if_true> if <condition> else <expression_if_false>
"""

num = 7
print(f"{num} is an \33[4m even \33[0m number.") if num % 2 == 0 else print(f"{num} is an \33[4m odd \33[0m number.")
```

**Testing Multiple Conditions with `in` and `not in`**

```python
"""
The code below checks if requested pizza toppings are available, adds them if they are, and notifies the customer if they aren't.
Finally, it confirms the pizza is ready.
"""

# List of available toppings in the restaurant
available_toppings = ["mushrooms", "pepperoni", "extra cheese", "green peppers", "olives"]

# List of toppings the customer wants
requested_toppings = ["pepperoni", "extra cheese", "chicken", "onions"]

# Checking each requested topping
for topping in requested_toppings:
    if topping in available_toppings:   # 'in' checks if a value exists inside a collection (list, dictionary, string, etc.).
        print(f"Adding {topping}.")
    else:
        print(f"Sorry, we are out of {topping} at the moment.")

# Final message
print("\nYour pizza is ready! Enjoy your meal!")
```

```python
available_toppings = ["mushrooms", "pepperoni", "extra cheese"]
requested_topping = "chicken"

if requested_topping not in available_toppings:
    print(f"Sorry, we don't have {requested_topping}.")
else:
    print(f"Adding {requested_topping}.")
```

**Checking if a list is empty**

```python
my_list = []

if not my_list:  # Equivalent to if my_list == [] or if len(my_list == 0)
    print("The list is empty.")
else:
    print("The list is not empty.")
```

**Logical Operators in If Statements**
You can combine multiple conditions using logical operators:

- `and` (both conditions must be True)
- `or` (at least one condition must be True)
- `not` (reverses the condition)

```python
x = 15

if x > 10 and x < 20:
    print("x is between 10 and 20")

if x < 10 or x > 20:
    print("x is not between 10 and 20")

if not (x == 10):
    print("x is not equal to 10")
```
